```python
nums = []

for i in range(10):
    n = int(input("Enter number: "))
    nums.append(n)

nums.sort()

print("Smallest =", nums[0])
print("Second Largest =", nums[-2])
print("Largest =", nums[-1])
```

```python
s = input("Enter string: ")

v = c = d = sp = 0

for ch in s:
    if ch in "aeiouAEIOU":
        v += 1
    elif ch.isalpha():
        c += 1
    elif ch.isdigit():
        d += 1
    else:
        sp += 1

print("Vowels =", v)
print("Consonants =", c)
print("Digits =", d)
print("Special Characters =", sp)
```

```python
num = int(input("Enter number: "))

temp = num
sum = 0

while temp > 0:
    digit = temp % 10
    sum = sum + digit ** 3
    temp = temp // 10

if sum == num:
    print("Armstrong Number")
else:
    print("Not Armstrong")
```

```python
students = (("Aman", 80), ("Riya", 95), ("Rahul", 88))

top_name = students[0][0]
top_marks = students[0][1]

for s in students:
    if s[1] > top_marks:
        top_marks = s[1]
        top_name = s[0]

print("Topper =", top_name)
print("Marks =", top_marks)
```

```python
sen = input("Enter sentence: ")

words = sen.split()

d = {}

for w in words:
    if w in d:
        d[w] += 1
    else:
        d[w] = 1

print(d)
```

```python
students = {
    "Aman": 70,
    "Riya": 90,
    "Rahul": 80
}

total = 0

for i in students:
    total += students[i]

avg = total / len(students)

print("Average =", avg)

for i in students:
    if students[i] > avg:
        print(i, students[i])
```

```python
products = {}

while True:
    print("1.Add")
    print("2.View")
    print("3.Exit")

    ch = int(input("Enter choice: "))

    if ch == 1:
        name = input("Product Name: ")
        price = input("Price: ")
        products[name] = price

    elif ch == 2:
        print(products)

    elif ch == 3:
        break
```

```python
a = int(input("Enter side 1: "))
b = int(input("Enter side 2: "))
c = int(input("Enter side 3: "))

if a == b and b == c:
    print("Equilateral")

elif a == b or b == c or a == c:
    print("Isosceles")

else:
    print("Scalene")
```
