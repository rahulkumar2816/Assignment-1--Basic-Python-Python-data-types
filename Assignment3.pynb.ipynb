```python
# 1. prime numbers from a list 

nums = [2, 5, 8, 11, 14, 17]

for n in nums:
    c = 0
    for i in range(1, n + 1):
        if n % i == 0:
            c += 1
    if c == 2:
        print(n)
```

```python
# 3. function from even count and odd sum
 
def check(nums):
    even = 0
    oddsum = 0

    for i in nums:
        if i % 2 == 0:
            even += 1
        else:
            oddsum += i

    return even, oddsum


a = [1, 2, 3, 4, 5]
e, o = check(a)

print("Even count =", e)
print("Odd sum =", o)
```

```python
# 4.Simple Interest using default arguments
 
def si(p, r=5, t=1):
    return (p * r * t) / 100


print(si(1000))
print(si(1000, 8, 2))
print(si(p=1000, r=10, t=3)) 
```

```python
#5 & 6. Student class

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def grade(self):
        if self.marks >= 80:
            return "A"
        elif self.marks >= 60:
            return "B"
        else:
            return "C"


s1 = Student("Ravi", 85)
s2 = Student("Anita", 65)

print(s1.name, s1.grade())
print(s2.name, s2.grade())
```

```python
# 7.  BankAccount class

class BankAccount:
    def __init__(self, bal):
        self.__bal = bal

    def deposit(self, amt):
        self.__bal += amt

    def withdraw(self, amt):
        if amt <= self.__bal:
            self.__bal -= amt
        else:
            print("Not enough balance")

    def show(self):
        print("Balance =", self.__bal)


a = BankAccount(5000)

a.deposit(1000)
a.withdraw(2000)
a.show()
```

```python
# 10. Read numbers from a file and find Total and Average
#Suppose numbers.txt contains:

try:
    f = open("numbers.txt", "r")

    total = 0
    count = 0

    for line in f:
        num = int(line)
        total += num
        count += 1

    avg = total / count

    print("Total =", total)
    print("Average =", avg)

    f.close()

except FileNotFoundError:
    print("File does not exist")
```

```python
# DMart Management System

products = {
    1: {"name": "Rice 5kg", "price": 350},
    2: {"name": "Sugar 1kg", "price": 45},
    3: {"name": "Milk 1L", "price": 60},
    4: {"name": "Oil 1L", "price": 150},
    5: {"name": "Soap", "price": 35}
}

customers = {}
sales = []


def add_customer():
    cid = len(customers) + 1
    name = input("Enter Customer Name: ")
    customers[cid] = {
        "name": name,
        "purchases": []
    }
    print(f"Customer Added Successfully. ID = {cid}")


def show_products():
    print("\n----- Product List -----")
    for pid, p in products.items():
        print(f"{pid}. {p['name']} - ₹{p['price']}")


def purchase_product():
    cid = int(input("Enter Customer ID: "))

    if cid not in customers:
        print("Customer Not Found!")
        return

    show_products()

    pid = int(input("Enter Product ID: "))
    qty = int(input("Enter Quantity: "))

    if pid not in products:
        print("Invalid Product!")
        return

    total = products[pid]["price"] * qty

    customers[cid]["purchases"].append({
        "product": products[pid]["name"],
        "qty": qty,
        "amount": total
    })

    sales.append({
        "customer": customers[cid]["name"],
        "product": products[pid]["name"],
        "qty": qty
    })

    print("\n----- BILL -----")
    print("Customer:", customers[cid]["name"])
    print("Product :", products[pid]["name"])
    print("Quantity:", qty)
    print("Total   : ₹", total)
    print("-----------------")


def customer_history():
    cid = int(input("Enter Customer ID: "))

    if cid not in customers:
        print("Customer Not Found!")
        return

    print("\nPurchase History")
    for item in customers[cid]["purchases"]:
        print(
            f"{item['product']} | Qty:{item['qty']} | Amount:₹{item['amount']}"
        )


def repeat_products():
    print("\n----- Repeat Products -----")

    for cid, customer in customers.items():
        count = {}

        for item in customer["purchases"]:
            product = item["product"]
            count[product] = count.get(product, 0) + 1

        for product, times in count.items():
            if times > 1:
                print(
                    f"{customer['name']} purchased {product} {times} times"
                )


def most_sold_products():
    sold = {}

    for sale in sales:
        sold[sale["product"]] = sold.get(
            sale["product"], 0
        ) + sale["qty"]

    print("\n----- Most Sold Products -----")

    sorted_products = sorted(
        sold.items(),
        key=lambda x: x[1],
        reverse=True
    )

    for product, qty in sorted_products:
        print(f"{product} : {qty} units sold")


while True:
    print("\n===== DMART MANAGEMENT SYSTEM =====")
    print("1. Add Customer")
    print("2. Show Products")
    print("3. Purchase Product")
    print("4. Customer History")
    print("5. Repeat Products")
    print("6. Most Sold Products")
    print("7. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_customer()

    elif choice == "2":
        show_products()

    elif choice == "3":
        purchase_product()

    elif choice == "4":
        customer_history()

    elif choice == "5":
        repeat_products()

    elif choice == "6":
        most_sold_products()

    elif choice == "7":
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```
